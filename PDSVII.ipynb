**Write a Python program to create a scatter plot of Age and Height of 5
students.**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
Age = np.array([18,20,21,19,22])
Height = np.array([5.5,5.6,5.4,5.3,5.7])

plt.scatter(Age, Height)

plt.xlabel("Age of student")
plt.ylabel("Height of student")
plt.show()
```

**b. Write a Python program to create a bar chart showing marks of 5
students.**

```python
name = np.array(["Amit", "Raj", "Suuny", "Kaushal"])
marks = np.array([55, 81, 51, 96])

plt.bar(name,marks)
plt.xlabel("name of student")
plt.ylabel("marks of student")
plt.show()
```

**Write a Python program to create a histogram showing the distribution of

**Write a Python program to create a pie chart showing the number of
